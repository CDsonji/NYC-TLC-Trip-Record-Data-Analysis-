```python
import duckdb

# Connect to an in-memory DuckDB instance
con = duckdb.connect(database=':memory:')

# Query all 12 monthly parquet files at once using a wildcard pattern (*)
query = """
SELECT 
    COUNT(*) AS total_trips,
    MIN(tpep_pickup_datetime) AS start_date,
    MAX(tpep_pickup_datetime) AS end_date,
    AVG(trip_distance) AS avg_distance,
    AVG(fare_amount) AS avg_fare
FROM './raw/yellow_tripdata_2025-*.parquet'
"""

print("Executing query across all 2025 data...")
df_summary = con.execute(query).df()
print(df_summary)
```
